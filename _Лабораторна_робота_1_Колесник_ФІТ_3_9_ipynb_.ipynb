### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
a = np.array([1, 2])
b = np.array([2, 1])

# Скалярний добуток та норми векторів
dot_product = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# Косинус кута
cos_theta = dot_product / (norm_a * norm_b)

# Кут у радіанах та градусах
angle_radians = np.arccos(cos_theta)
angle_degrees = np.degrees(angle_radians)

print("Косинус кута:", cos_theta)
print("Кут у градусах:", angle_degrees)
````

### Завдання 27
Дано матриці:

```text
A = [[1, 2],
     [3, 4]]

B = [[2, 0],
     [1, 2]]
```

Обчисліть:
1. поелементний добуток `A * B`;
2. матричний добуток `A @ B`.

Поясніть у коментарі до коду, чим відрізняються ці операції.

````python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[2, 0],
              [1, 2]])

# Поелементний добуток (Hadamard product):
# Кожен елемент матриці A множиться на відповідний елемент матриці B (a_ij * b_ij)
elementwise = A * B

# Матричний добуток (Matrix multiplication):
# Обчислюється як скалярний добуток рядків матриці A та стовпців матриці B
matrix_mult = A @ B

print("Поелементний добуток (A * B):\n", elementwise)
print("\nМатричний добуток (A @ B):\n", matrix_mult)
````

### Завдання 28
Для матриці:

```text
A = [[4, 7],
     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
A = np.array([[4, 7],
              [2, 6]])

# Визначник
det_A = np.linalg.det(A)

# Обернена матриця
inv_A = np.linalg.inv(A)

# Перевірка (A @ A_inv має дорівнювати одиничній матриці I)
check = A @ inv_A

print("Визначник:", det_A)
print("\nОбернена матриця:\n", inv_A)
print("\nПеревірка (A @ A^-1):\n", check)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]])

# Середнє та стандартне відхилення для кожного стовпця (axis=0)
mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

# Z-score нормалізація за допомогою broadcasting
normalized_A = (A - mean) / std

print(normalized_A)
````

````python
np.random.seed(42)

# Моделювання оцінок 30 студентів
grades = np.random.randint(0, 101, size=30)

# Статистичні показники
mean_grade = np.mean(grades)
median_grade = np.median(grades)
max_grade = np.max(grades)
min_grade = np.min(grades)

# Кількісні та відсоткові показники
passed_count = np.sum(grades >= 60)
top_percentage = np.mean(grades >= 90) * 100

print("Оцінки студентів:\n", grades)
print("\nСередня оцінка:", mean_grade)
print("Медіана:", median_grade)
print("Найвища оцінка:", max_grade)
print("Найнижча оцінка:", min_grade)
print("Кількість студентів з оцінкою >= 60:", passed_count)
print(f"Відсоток студентів з оцінкою >= 90: {top_percentage:.2f}%")
````

Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
sales = np.array([[120, 150, 170, 160],
                  [80,  90,  100, 110],
                  [200, 180, 210, 230]])

# Загальний обсяг продажів кожного товару (сума по рядочках)
total_per_product = np.sum(sales, axis=1)

# Загальний обсяг продажів у кожному місяці (сума по стовпчиках)
total_per_month = np.sum(sales, axis=0)

# Товар із найбільшими сумарними продажами (індекс + 1)
best_product_idx = np.argmax(total_per_product) + 1

# Місяць із найбільшим загальним обсягом продажів (індекс + 1)
best_month_idx = np.argmax(total_per_month) + 1

# Середній обсяг продажів кожного товару
mean_per_product = np.mean(sales, axis=1)

print("Загальні продажі за товарами (1, 2, 3):", total_per_product)
print("Загальні продажі за місяцями (1, 2, 3, 4):", total_per_month)
print("Товар з найбільшими продажами:", best_product_idx)
print("Місяць з найбільшими продажами:", best_month_idx)
print("Середні продажі за товарами:", mean_per_product)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
np.random.seed(42)

# Створення матриці 5x5 із випадкових цілих чисел від 1 до 20
A = np.random.randint(1, 21, size=(5, 5))

# 1. Сума головної діагоналі
main_diag_sum = np.trace(A)

# 2. Сума побічної діагоналі
sec_diag_sum = np.trace(np.fliplr(A))

# 3. Максимальний елемент кожного рядка
max_per_row = np.max(A, axis=1)

# 4. Мінімальний елемент кожного стовпця
min_per_col = np.min(A, axis=0)

# 5. Заміна елементів, менших за середнє значення всей матриці, на 0
A_modified = A.copy()
A_modified[A_modified < np.mean(A)] = 0

print("Початкова матриця A:\n", A)
print("\nСума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", sec_diag_sum)
print("Максимуми по рядках:", max_per_row)
print("Мінімуми по стовпцях:", min_per_col)
print("\nМатриця після заміни елементів < mean на 0:\n", A_modified)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
np.random.seed(42)

# Створення матриці 4x4 із довільних цілих чисел
A = np.random.randint(1, 10, size=(4, 4))

# Норма кожного рядка (axis=1), keepdims=True для збереження форми (4, 1) та сумісності з broadcasting
row_norms = np.linalg.norm(A, axis=1, keepdims=True)

# Нормалізація кожного рядка до одиничної довжини
A_normalized = A / row_norms

print("Початкова матриця:\n", A)
print("\nНорми рядків:\n", row_norms)
print("\nНормалізована матриця:\n", A_normalized)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
import numpy as np

# Матриця коефіцієнтів A та вектор правих частин b
A = np.array([[1,  2,  1],
              [2,  1, -1],
              [3, -1,  2]])

b = np.array([9, 8, 3])

# Розв'язок системи Ax = b
x = np.linalg.solve(A, b)

# Перевірка матричним множенням (A @ x)
b_check = A @ x

print("Розв'язок (x, y, z):", x)
print("Результат A @ x:", b_check)
print("Результат збігається з вектором b:", np.allclose(b_check, b))
````
